```python
import requests
import nba_api
import polars as pl
import io
```

```python
from nba_api.stats.static import *
```

## PLAYERS COLLECT 

```python
players_list = players.get_players()
```

```python
PLAYERS_SCHEMA = {
    'id': pl.Utf8,
    'full_name' : pl.Utf8, 
    'first_name' : pl.Utf8,
    'last_name' : pl.Utf8, 
    'is_active' : pl.Boolean
}
```

```python
playes_lf = pl.DataFrame(players_list , schema = PLAYERS_SCHEMA).lazy().collect()
```

## TEAMS COLLECT

```python
teams_list = teams.get_teams()
print(teams_list[0].keys())
```

```python
TEAMS_SCHEMA = {
    'id': pl.Utf8,
    'full_name' : pl.Utf8, 
    'abbreviation' : pl.Utf8,
    'nickname' : pl.Utf8, 
    'city' : pl.Utf8,
    'state' : pl.Utf8,
    'year_founded' : pl.Int64
}
```

```python
teams_lf = pl.DataFrame(teams_list , schema = TEAMS_SCHEMA).lazy().collect()
```

## SAVE

```python
teams_lf.write_parquet("teams_raw.parquet")
playes_lf.write_parquet("players_raw.parquet")
```
